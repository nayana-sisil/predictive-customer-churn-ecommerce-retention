## 1.0 BUSINESS CONTEXT & PROBLEM STATEMENT

### 1.1 Platform Growth Challenge

The e commerce platform has experienced significant user growth, reaching millions of active users. However, this growth masks a critical retention problem: a growing number of customers are becoming inactive, directly impacting long term revenue and customer lifetime value.

In today's highly competitive e commerce market, customers can easily switch between platforms. Industry research shows that **acquiring a new customer costs 5-7× more than retaining an existing one**, making customer retention as strategically important as customer acquisition.

### 1.2 Current Retention Strategy Limitations

The platform has implemented standard retention mechanisms:
- Loyalty programs
- Frequent promotional campaigns
- Periodic discount offers

However, these approaches show **limited effectiveness** due to:
- **Lack of Personalization** - Generic offers fail to address individual customer behavior and preferences
- **Poor Targeting** - Inability to distinguish between at-risk customers and already loyal customers
- **Resource Waste** - Promotions reach customers who don't need additional incentives
- **Missed Opportunities** - Truly at-risk customers are not identified early enough for intervention

### 1.3 Problem Hypothesis

**"The platform's inability to distinguish between customers at risk of churn and those who are already loyal results in indiscriminate promotion distribution. This approach wastes resources on customers who don't require additional incentives while failing to effectively target those who are truly at risk of churning."**

### 1.4 Primary Stakeholder

**Head of Customer Experience and Marketing Optimization**

This role serves as the main stakeholder because it provides:
- **Comprehensive Oversight** - Manages retention initiatives, monitors behavior, and oversees marketing budget allocation

- **Data-Driven Decision Making** - Has access to customer analytics for informed decisions

- **Direct Impact** - Directly affects KPIs such as retention rate, CLV, and marketing ROI

### 2.2 Evaluation Metric: F2-Score (β=2)/FBeta Score

**Why F2-Score?**

The F2-score is selected as the primary evaluation metric because it prioritizes **recall over precision** by giving twice the weight to recall. This aligns perfectly with the business objective.

**Business Logic:**

| Scenario | Business Impact | Cost |
|----------|----------------|------|
| **False Negative** (Miss a churner) | Customer lost, revenue permanently gone | **Very High** (RS 1,200 CLV) |
| **False Positive** (Offer to loyal customer) | Unnecessary promotion given | **Low** (RS 50 offer) |

**Key Reasons:**
1. The cost of losing a customer (false negative) is **24× higher** than providing an unnecessary promotion (false positive)
2. Early identification enables timely preventive actions
3. False positives don't harm loyal customers retention benefits are generally positive
4. Aggressive churn capture aligns with business need to minimize customer loss

## 3.0 DATA OVERVIEW & PREPARATION

### 3.1 Dataset Source

**Origin:** HackerEarth Machine Learning Challenge (2021)  
**Challenge Title:** "How NOT to lose a customer in 10 days"  
**Total Records:** 3,941 customers  
**Features:** 10 input features + 1 target variable (Churn)

### 3.2 Dataset Dictionary

| Feature | Type | Description |
|---------|------|-------------|
| Tenure | Float | Tenure of customer with the company (months) |
| WarehouseToHome | Float | Distance from warehouse to customer's home (km) |
| NumberOfDeviceRegistered | Integer | Total devices registered by customer |
| PreferedOrderCat | Categorical | Preferred order category in last month |
| SatisfactionScore | Integer | Customer satisfaction score (1-5 scale) |
| MaritalStatus | Categorical | Marital status of customer |
| NumberOfAddress | Integer | Total addresses added by customer |
| Complain | Binary | Whether complaint raised in last month (0/1) |
| DaySinceLastOrder | Float | Days since customer's last order |
| CashbackAmount | Float | Average cashback received in last month (RS) |
| **Churn** | Binary | **Target variable** - Customer churned (1) or not (0) |

### 3.3 Initial Data Quality Assessment

**Missing Values:**
| Feature | Missing Count | Missing % |
|---------|---------------|-----------|
| DaySinceLastOrder | 213 | 5.4% |
| Tenure | 194 | 4.9% |
| WarehouseToHome | 169 | 4.3% |

**Observation:** Missing values show a **Missing At Random (MAR)** pattern, independent of the target variable. Advanced imputation methods were required.

**Duplicate Records:**
- **671 duplicate rows** identified (17% of dataset)
- **Action:** Removed duplicates, keeping first occurrence

**Class Imbalance:**
| Churn Status | Count | Percentage |
|--------------|-------|------------|
| Not Churned (0) | 3,267 | 82.9% |
| Churned (1) | 674 | 17.1% |

**Critical Finding:** Severe class imbalance (4.8:1 ratio) requiring specialized handling techniques to prevent model bias toward majority class.

### 3.4 Data Cleaning Steps

1. **Duplicate Removal:** 671 duplicate records removed, resulting in 3,270 unique records

2. **Categorical Standardization:**
   - Merged 'Mobile' and 'Mobile Phone' categories in PreferedOrderCat
   - Result: Consolidated from 6 to 5 meaningful categories

3. **Outlier Analysis:**
   - Outliers detected in Tenure, WarehouseToHome, DaySinceLastOrder, CashbackAmount
   - **Decision:** Retained outliers as they represent valid customer behavior patterns
   - Applied robust scaling to handle outliers without removing valuable information

4. **Missing Value Strategy:**
   - Used **KNN Imputation (k=10)** to preserve data patterns
   - Avoided simple mean/median filling to maintain data integrity

### 3.5 Train Test Split

**Strategy:** Stratified 80-20 split to maintain class distribution

**Results:**
- Training Set: 2,616 samples (83.7% non-churn, 16.3% churn)
- Test Set: 654 samples (83.6% non-churn, 16.4% churn)
- **Validation:** Perfect stratification achieved

## 4.0 METHODOLOGY & ANALYTICAL APPROACH

### 4.1 End to End Pipeline

```
[Raw Data]
    ↓
[Data Cleaning]
  • Remove duplicates
  • Standardize categories
  • Retain meaningful outliers
    ↓
[Train-Test Split (80-20, Stratified)]
    ↓
[Preprocessing Pipeline]
  • KNN Imputation (k=10)
  • Power Transformation (Yeo-Johnson)
  • Robust Scaling
  • One-Hot Encoding
    ↓
[Handle Class Imbalance]
  • Tested: SMOTE, BorderlineSMOTE, RandomOverSampler
  • Tested: NearMiss, TomekLinks, ClusterCentroids
  • Selected: RandomOverSampler (best performance)
    ↓
[Model Training & Selection]
  • Cross Validation: 5-Fold Stratified
  • Optimization: RandomizedSearchCV (30 iterations)
  • Metric: F2-Score (β=2)
    ↓
[Model Evaluation]
  • Confusion Matrix
  • ROC AUC, PR-AUC
  • Feature Importance
  • SHAP Analysis
    ↓
[Business Impact Analysis]
  • Customer profiling
  • Financial projections
  • ROI calculation
    ↓
[Deployment Recommendations]
```

### 4.2 Preprocessing Strategy

**Numerical Features Pipeline:**
1. **KNN Imputation (k=10)** - Preserves multivariate patterns in missing data
2. **Power Transformation (Yeo-Johnson)** - Normalizes skewed distributions
3. **Robust Scaling** - Centers and scales data while being resistant to outliers

**Categorical Features:**
- **One Hot Encoding** with drop='if_binary' to avoid multicollinearity

**Why This Approach?**
- KNN imputation outperformed simple mean/median filling by preserving feature relationships
- Power Transformation improved model performance for linear models by 1.7% (F2-score)
- Robust Scaling handled outliers better than Standard Scaling, improving stability

### 4.3 Class Imbalance Handling

**Methods Evaluated:**

| Technique | Type | Mechanism | F2-Score Result |
|-----------|------|-----------|----------------|
| **RandomOverSampler** | Oversample | Duplicate minority samples | **0.892** ✓ |
| SMOTE | Oversample | Synthetic sample generation | 0.872 |
| BorderlineSMOTE | Oversample | Focus on borderline cases | 0.861 |
| NearMiss | Undersample | Remove majority samples | 0.854 |
| TomekLinks | Hybrid | Remove boundary samples | 0.855 |
| ClusterCentroids | Undersample | Replace with centroids | 0.488 |
| Class Weight='balanced' | Algorithmic | Cost-sensitive learning | 0.881 |

**Selection:** RandomOverSampler chosen for final model as it achieved highest F2 score while maintaining training stability.

### 5.2 Confusion Matrix Analysis

**Test Set Results (n=654):**

|  | Predicted Not Churn | Predicted Churn |
|---|---------------------|-----------------|
| **Actually Not Churn** | 505 (TN) | 42 (FP) |
| **Actually Churn** | 12 (FN) | 95 (TP) |

**Interpretation:**
- **True Positives (95):** Successfully identified churners - these customers receive targeted interventions
- **False Negatives (12):** Missed churners - represents 11.2% of actual churners (acceptable given business trade-offs)
- **False Positives (42):** Loyal customers receiving unnecessary offers - low cost impact (RS 2,100 total)
- **True Negatives (505):** Correctly identified loyal customers - no intervention needed

**Key Insight:** The model misses only 12 out of 107 churners, achieving the business objective of maximizing churn capture while maintaining acceptable precision.

### 6.3 High Risk Customer Profile

**Demographic & Behavioral Characteristics:**

**Profile of a Typical Churner:**
- **Tenure:** 0-3 months (new customer)
- **Complain Status:** Has raised complaint in last 30 days
- **Cashback Received:** Below RS 160 per month
- **Warehouse Distance:** >15 km from nearest warehouse
- **Recent Activity:** Made purchase within last 3-7 days
- **Satisfaction Score:** 2-3 (out of 5)
- **Device Registration:** 3-4 devices (moderate engagement)

**Behavioral Patterns Leading to Churn:**

1. **"Disappointed Early Adopter"**
   - Signs up with high expectations
   - Experiences service issue or complaint within first month
   - Issue not resolved satisfactorily
   - Churns before month 3

2. **"Delivery-Frustrated Customer"**
   - Lives >20km from warehouse
   - Experiences delayed or inconsistent deliveries
   - Lower satisfaction with logistics
   - Gradually disengages over 2-4 months

3. **"Reward-Sensitive Shopper"**
   - Attracted by initial promotions
   - Receives below average cashback
   - Finds better deals on competitor platforms
   - Churns within first 6 months

## 10.0 LIMITATIONS & FUTURE ENHANCEMENTS

### 10.1 Current Model Limitations

**Data Limitations:**
1. **Historical Bias:** Model trained on past behavior; may not capture emerging churn patterns
2. **Feature Coverage:** Limited to 10 input features; potentially missing behavioral signals
3. **Temporal Dynamics:** Does not account for seasonality or temporal trends
4. **External Factors:** Cannot capture competitor actions or market dynamics

**Technical Limitations:**
1. **Prediction Window:** Current model predicts churn without specific timeframe
2. **Interpretability Trade-off:** While SHAP provides insights, complex interactions remain opaque
3. **Threshold Optimization:** Fixed classification threshold; could benefit from dynamic adjustment
4. **Sample Size:** Training on 3,270 records; larger datasets could improve generalization

**Business Limitations:**
1. **Intervention Assumptions:** 70% retention success rate is estimated, not validated
2. **Cost Assumptions:** Fixed offer cost may not reflect actual campaign complexity
3. **Behavioral Response:** Model doesn't account for customer response to interventions

### Appendix B: Detailed Model Parameters

**Final LightGBM Configuration:**

```python
LGBMClassifier(
    random_state=42,
    class_weight='balanced',
    verbose=-1,
    subsample=0.8,
    reg_lambda=0.5,
    reg_alpha=0.5,
    num_leaves=50,
    n_estimators=100,
    min_child_samples=20,
    max_depth=20,
    learning_rate=0.05,
    colsample_bytree=0.6
)
```

**Preprocessing Pipeline:**
- KNN Imputation: k=10 neighbors
- Power Transformation: Yeo Johnson method
- Robust Scaling: IQR based centering and scaling
- One-Hot Encoding: Binary features dropped to avoid multicollinearity

**Resampling:**
- RandomOverSampler with random_state=42
- Balanced training set: 2,188 non-churn + 2,188 churn
